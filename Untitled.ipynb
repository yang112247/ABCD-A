```python
from ultralytics import YOLO
import cv2

# 모델 로드 (COCO 데이터셋으로 사전 학습됨)
model = YOLO("yolov8n.pt")

# 이미지 불러오기
image_path = "cat_image.jpg"
image = cv2.imread(image_path)

# 추론 실행
results = model(image)

# 감지된 객체 결과 확인
results[0].show()

# 고양이 클래스: COCO 기준으로 class ID 15가 고양이
cat_count = 0
for box in results[0].boxes:
    if int(box.cls[0]) == 15:
        cat_count += 1

print(f"감지된 고양이 수: {cat_count}")
```

```python
from ultralytics import YOLO
import cv2
import sys # Import sys to exit if image loading fails

# 모델 로드 (COCO 데이터셋으로 사전 학습됨)
model = YOLO("yolov8n.pt")

# 이미지 불러오기
image_path = "cat_image.jpg"
image = cv2.imread(image_path)

# 이미지 로드 성공 여부 확인
if image is None:
    print(f"오류: 이미지를 로드할 수 없습니다. 경로를 확인하세요: {image_path}")
    sys.exit() # Exit the script if the image is not loaded

# 추론 실행
results = model(image)

# 감지된 객체 결과 확인
results[0].show()

# 고양이 클래스: COCO 기준으로 class ID 15가 고양이
cat_count = 0
for box in results[0].boxes:
    if int(box.cls[0]) == 15:
        cat_count += 1

print(f"감지된 고양이 수: {cat_count}")

# %%
# This cell depends on the 'image' variable from the previous cell.
# Since we added a check and potential exit in the previous cell,
# this cell will only execute if the image was loaded successfully.
for box in results[0].boxes:
    cls_id = int(box.cls[0])
    if cls_id == 15:  # 고양이
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(image, "Cat", (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

cv2.imwrite("detected_cats.jpg", image)
cv2.imshow("Cats", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
